# 01 – Daten laden und Stichprobe ziehen

Dieses Notebook lädt die Consumer Complaint Database, beschränkt sie auf Beschwerden mit Text, entfernt Duplikate und zieht eine Stichprobe für die Analyse.

**Eingabe:** 
`data/rows.csv` (nicht im Repository)
**Ausgabe:** 
`data/sample_20000.csv` (lokal) und `sample/sample_500.csv` (im Repository)

```python
import pandas as pd
from pathlib import Path
```

```python

df = pd.read_csv("data/rows.csv")
print(df.shape)
print(df.columns.tolist())
df.head()
```

## Beschwerden mit Text und Duplikate

Der Datensatz hat 1.282.355 Beschwerden mit 18 Spalten. Nur 383.564 davon  enthalten einen Beschwerdetext, vermutlich, weil Verbraucher der Veröffentlichung zustimmen müssen. Für die Analyse behalte ich nur diese Zeilen und die Spalten `Complaint ID`, `Product`, `Issue` und den Text.

Danach entferne ich Duplikate anhand des Textes: 366.945 Beschwerden bleiben, 16.619  fallen weg.

```python
col = "Consumer complaint narrative"
print(df[col].notna().sum(), "von", len(df))
```

```python
df_text = df[["Complaint ID", "Product", "Issue", col]].dropna(subset=[col])
print(df_text.shape)

df_text = df_text.drop_duplicates(subset=[col])
print(df_text.shape)
```

## Stichprobe

Aus den 366.945 Beschwerden ziehe ich 20.000 zufällig. Die Zahl ist ein Kompromiss: Die Verarbeitung mit spaCy dauert damit etwa vier Minuten. `random_state=42` macht die Auswahl reproduzierbar.

Die Stichprobe liegt in `data/` und damit nicht im Repository. Die ersten 500 Zeilen speichere ich zusätzlich in `sample/sample_500.csv`, damit der Code auch ohne den großen Datensatz ausprobiert werden kann.

```python
sample = df_text.sample(n=20000, random_state=42)
sample.to_csv("data/sample_20000.csv", index=False)
```

```python
# Prüfung auf Darstellung der Texte
for t in df_text[col].head(3):
    print(t)
    print("---")
```
